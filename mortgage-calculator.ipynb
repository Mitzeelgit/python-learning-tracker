```python
def mortgagecalc():
    #principle amount
    while True:
        try:
            p = float(input("What is the initial loan amount borrowed"))
            if p > 0:
                break
            else:
                print("Value must be greater than 0")
        except ValueError:
            print("Enter a valid float")

    #annual rate
    while True:
        try:
            r = float(input("At what rate have you taken interest"))
            if r > 0:
                r = r/100
                break
            else:
                print("Value must be greater than 0")
        except ValueError:
            print("Enter a valid float")

    #years
    while True:
        try:
            t = float(input("For how long have you taken the loan in years"))
            if t > 0:
                break
            else:
                print("Value must be greater than 0")
        except ValueError:
            print("Enter a valid float")

    #compounding frequency
    while True:
        try:
            f = float(input("How many payments are you make per year"))
            if f > 0:
                break
            else:
                print("Value must be greater than 0")
        except ValueError:
            print("Enter a valid float")
   
    #compounding type with input validation

    compounding_options = ['monthly' , 'weekly' , 'daily' , 'continous']
    compounding_choosen = 'wrong'

    while compounding_choosen not in compounding_options:
        compouding_choosen = input("How frequently are you choosing to compound (monthly , weekly , daily , continous)?").lower()

        if compounding_choosen not in compounding_options:
            print("Choose a valid frequency between (monthly , weekly , daily , continous)")

    #setting value of n for each compounding type
    if compounding_choosen == 'monthly'.lower():
        n = 12
    elif compounding_choosen == 'weekly'.lower():
        n = 52
    elif compounding_choosen == 'daily'.lower():
        n = 365
    else:
        if compounding_choosen == 'continous'.lower():
            n = None
    #calculating INTEREST

    #for discrete compounding (monthly , weekly , daily)
    # e = 2.718 
    if n == 12 or n == 52 or n == 365:

        i = (1 + (r/n))**(n/f) - 1
    else:
        i = 2.718**(r/f) - 1

    #total payments

    total_payments = t*f

    #periodic payment

    monthly_payment = p * (i*(1+i)**total_payments) / ((1+i)**total_payments - 1)

    total_paid = monthly_payment * total_payments

    total_interest = total_paid - p

    #summary

    print("-----------MORTGAGE SUMMARY-----------")
    print("Payments per Period: ${}".format(monthly_payment))
    print("Total Payments made: {}".format(total_payments))
    print("Total Amount paid: ${}".format(total_paid))
    print("Total interest paid: ${}".format(total_interest))
```

```python
mortgagecalc()
```
