```python
import random

words_list = ['water','cat','bill','python','juice','ladder','goose','watch','clock']

word_choosen = random.choice(words_list)

no_of_blanks = len(word_choosen)

guessed_letters = []

no_of_lives = 6

hangman_stages = [
    # Stage 0: 0 lives remaining 
    """
       +---+
       |   |
       O   |
      /|\\  |
      / \\  |
           |
    =========
    """,
    # Stage 1: 1 life remaining
    """
       +---+
       |   |
       O   |
      /|\\  |
      /    |
           |
    =========
    """,
    # Stage 2: 2 lives remaining
    """
       +---+
       |   |
       O   |
      /|\\  |
           |
           |
    =========
    """,
    # Stage 3: 3 lives remaining
    """
       +---+
       |   |
       O   |
      /|   |
           |
           |
    =========
    """,
    # Stage 4: 4 lives remaining
    """
       +---+
       |   |
       O   |
       |   |
           |
           |
    =========
    """,
    # Stage 5: 5 lives remaining
    """
       +---+
       |   |
       O   |
           |
           |
           |
    =========
    """,
    # Stage 6: 6 lives remaining (Start)
    """
       +---+
       |   |
           |
           |
           |
           |
    =========
    """
]

print('WELCOME TO :')

print(''' ''')

while no_of_lives > 0:

    print(hangman_stages[no_of_lives])

    print('Lives left {}'.format(no_of_lives))

    display = ""

    for letter in word_choosen:
        if letter in guessed_letters:
            display += letter + " "
        else:
            display += "_ "
    
    print('Word : {}'.format(display))

    if '_' not in display:
        print('Congratulations ! You won')
        break

    guess = input("Guess a letter: ").lower()

    if len(guess) > 1:
        print('Enter a single letter')

    if guess in guessed_letters:
        print("You already guessed that letter!")
    elif guess in word_choosen:
        print("Nice! That letter is in the word.")
        guessed_letters.append(guess)
    else:
        print("Sorry, that letter isn't in the word.")
        guessed_letters.append(guess)
        no_of_lives -= 1

    if no_of_lives == 0:
        print(hangman_stages[0])
        print('Game over , the word was {}'.format(word_choosen))
```
