```python
def findpi():

    while True:
        try:
            till_no = int(input("Till what decimal place do you want the value of pi?"))
        except:
            print("Please enter a valid integer")
            continue
        else:
            final_val = (round(22/7 , till_no))
            break
    print("Value of pi till {} is {}".format(till_no,final_val))
            
```
