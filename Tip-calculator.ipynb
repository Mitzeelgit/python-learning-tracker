```python
print('Welcome to the tip calculator')

total_bill = float(input('What was the total bill?'))

tip = int(input('What percent of the bill would you like to tip 10% , 12% or 15%'))

people = int(input('How many people to split the bill?'))

pp_bill = (total_bill/people) +  (total_bill * (tip/100))/people

print(pp_bill)
```
