```python
def tilecostcalc():

    while True:
        
        try:
            width = float(input("What is the width of your room"))
            height = float(input("What is the height of your room"))
            cost = float(input("What is the price per sq meter of your tile"))
            break
        except:
            print("Please enter a valid float")
            continue
    if width > 0 and height > 0 and cost > 0:
        area = width*height       
        price = (area*cost)
        print("The total cost for your room tile plan is ${}".format(price))
    elif width == 0 or height == 0 or cost ==0:
        print("Values cannot be zero")
    else:
        print("Values cannot be negative")
            
```
