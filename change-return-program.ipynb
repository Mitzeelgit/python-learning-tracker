```python
import math
def changereturn():

    #ask for cost
    while True:
        try:
            cost = float(input("Enter the cost of the item"))
            if cost > 0:
                break
            else:
                print("Cost can not be negative")
        except ValueError:
            print("Enter a valid float")

    #amount given
    while True:
        try:
            given = float(input("Enter the amount you paid of the item"))
            if given >= cost:
                break
            else:
                print("Cost can not be more than given amount")
        except ValueError:
            print("Enter a valid float")

    #calculating change
    change = given - cost
    if change > 0:
        pass       
    else:
        print("Cost cannot be more than amount given")

    #various values
    penny = 0.01
    nickels = 0.05
    dime = 0.1
    quarter = 0.25

    no_of_penny = change/penny

    no_of_nickels = change/nickels

    no_of_dime = change/dime

    no_of_quarter = change/quarter

    print("----------CHANGE CALCULATOR---------------------")
    print("No.of pennies needed: {}".format(math.ceil(no_of_penny)))
    print("No.of nickels needed: {}".format(math.ceil(no_of_nickels)))
    print("No.of dimes needed: {}".format(math.ceil(no_of_dime)))
    print("No.of quarters needed: {}".format(math.ceil(no_of_quarter)))
```
