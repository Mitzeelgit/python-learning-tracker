```python
def prime_list():

    while True:
        try:
            till = int(input("Till what number do you want prime numbers"))
        except:
            print("Please enter a valid integer")
            continue
        else:
            sequence = []
            for i in range(2,till+1):
                is_prime = True
                for j in range(2,i):
                    if i%j == 0:
                        is_prime = False
                        break
                if is_prime:
                    sequence.append(i)
            break
                   
    print("Prime numbers until {} are {}".format(till,sequence))
                    
```

```python
prime_list()
```
