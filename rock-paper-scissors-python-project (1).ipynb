```python
#function to print rock
def rock():
    print('''             _______
         ---'   ____)
               (_____)
               (_____)
               (____)
         ---.__(___)''')
```

```python
rock()
```

```python
#function to print paper
def paper():
    print('''    _______
---'   ____)____
          ______)
          _______)
         _______)
---.__________)''')
```

```python
paper()
```

```python
#function to print scissors
def scissors():
    print('''    _______
---'   ____)____
          ______)
       __________)
      (____)
---.__(___)''')
```

```python
scissors()
```

```python
#random choice of computer
import random

def computer_choice():

    comp_choice = random.randint(0,2)

    if comp_choice == 0:
        print('Computer CHOOSE :')
        rock()
        return 0
    elif comp_choice == 1:
        print('Computer CHOOSE :')
        paper()
        return 1
    else:
        print('Computer CHOOSE :')
        scissors()
        return 2
```

```python
#ASKS FOR PLAYER INPUT AND VALIDATES IT
def player_input():

    player_choice = []

    player_options = [0,1,2]

    while player_choice not in player_options:
        player_choice = int(input('What do you choose? type 0 for rock , 1 for paper , 2 for scissors[MAKE SURE TO TYPE AN INTEGER]:'))

        if player_choice not in player_options:
            print('Choose a valid number : 0 , 1 or 2')
    return player_choice

        
```

```python
#FINAL CODE
#____  _   _ _   _   _____ _   _ ___ ____  
#|  _ \| | | | \ | | |_   _| | | |_ _/ ___| 
#| |_) | | | |  \| |   | | | |_| || |\___ \ 
#|  _ <| |_| | |\  |   | | |  _  || | ___) |
#|_| \_\\___/|_| \_|   |_| |_| |_|___|____/
print('Welcome to Rock , Paper , Scissors')
print('What do you choose? type 0 for rock , 1 for paper , 2 for scissors')
comp_choice = computer_choice()

player_choice = player_input()

if player_choice == 0 and comp_choice == 0:
    print('Player CHOOSE:')
    rock()
    print('Both player and computer choose ROCK and its a DRAW!')
elif player_choice == 0 and comp_choice == 1:
    print('Player CHOOSE:')
    rock()
    print('Computer choose PAPER and WON!')
elif player_choice == 0 and comp_choice == 2:
    print('Player CHOOSE:')
    rock()
    print('Player choose ROCK and WON!')
elif player_choice == 1 and comp_choice == 1:
    print('Player CHOOSE:')
    paper()
    print('Both player and computer choose PAPER and its a DRAW!')
elif player_choice == 1 and comp_choice == 2:
    print('Player CHOOSE:')
    paper()
    print('Computer choose SCISSORS and WON!')
elif player_choice == 2 and comp_choice == 2:
    print('Player CHOOSE:')
    scissors()
    print('Both player and computer choose SCISSORS and its a DRAW!')
elif player_choice == 2 and comp_choice == 0:
    print('Player CHOOSE:')
    scissors()
    print('Computer choose ROCK and WON!')
elif player_choice == 2 and comp_choice == 1:
    print('Player CHOOSE:')
    scissors()
    print('Player choose SCISSORS and WON!')
elif player_choice == 1 and comp_choice == 0:
    print('Player CHOOSE:')
    paper()
    print('Player choose PAPER and WON!')
```
